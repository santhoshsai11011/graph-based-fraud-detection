```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

edges = pd.read_csv(
    "../data/raw/elliptic_txs_edgelist.csv"
)
```

```python
G = nx.from_pandas_edgelist(
    edges,
    source="txId1",
    target="txId2",
    create_using=nx.DiGraph()
)
```

```python
degrees = [
    degree
    for node, degree
    in G.degree()
]

plt.figure(figsize=(10,6))

plt.hist(
    degrees,
    bins=50
)

plt.title(
    "Degree Distribution"
)

plt.xlabel("Degree")
plt.ylabel("Frequency")

plt.show()
```

```python
degree_dict = dict(
    G.degree()
)

top_nodes = sorted(
    degree_dict.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

top_nodes
```
